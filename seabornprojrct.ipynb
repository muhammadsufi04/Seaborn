```python
import matplotlib.pyplot as plt

import seaborn as sns

df=sns.load_dataset("penguins")

sns.lineplot(x="bill_length_mm",y='flipper_length_mm',data=df,hue='sex')

plt.show()
```

```python
sns.set(style="dark")
sns.barplot(x="island",y='bill_length_mm',data=df,hue="sex")
plt.show()
```

```python
sns.displot(df['flipper_length_mm'],kde=True,rug=True)
plt.show()
```

```python

sns.scatterplot(x='bill_length_mm',y="bill_length_mm",data=df,hue="sex",style='sex',size='sex',sizes=(100,40))
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

df=sns.load_dataset("tips")

sns.countplot(x="sex",data=df)
plt.show()
```

```python

sns.violinplot(x="day",y='total_bill',data=df,hue='sex',split=True)
plt.show()
```

```python
sns.stripplot(x='day',y='total_bill',data=df,edgecolor='r',size=5,marker='<')
plt.show()
```

```python

sns.boxplot(x='day',y='total_bill',data=df,hue="sex",showmeans=True,linewidth=3,palette='plasma')
plt.show()
```

```python
sns.catplot(x='tip',y="size",data=df,hue='sex',kind='bar')
plt.show()
```

```python
sns.set_style('dark')
sns.set_context('poster',font_scale=1)
sns.barplot(x='day',y='total_bill',data=df)
sns.despine()
plt.show()
```

```python
fg=sns.FacetGrid(df,col='sex')
fg.map(plt.scatter,'total_bill','tip')
plt.show()
```
